### Example 1 : Write a function that calculates geometric and arithmetic mean

```python
import numpy as np

def ArithGeoMean( a , b , mode ) :
    if mode == "arithmatic" :
        out = (a + b)/2
    elif mode == "geometric" :
        out = ( a * b ) ** 0.5
    return out

def ArithGeoMean2( a , b = 2 , mode = "geometric" ) :
    if mode == "arithmatic" :
        out = (a + b)/2
    elif mode == "geometric" :
        out = ( a * b ) ** 0.5
    return out

Ans1 = ArithGeoMean2(4 , 100 , 'geometric')
print(Ans1)
Ans2 = ArithGeoMean2(4)
print(Ans2)
```

### Example 2 : Geometric mean with tuple return

```python
import numpy as np

def GeoAvg ( Inputs ) :
    Result = 1
    for Item in Inputs :
        Result *= Item
    return Result**(1/len(Inputs)), len(Inputs)

Q1 , Q2 = GeoAvg( [1 , 2 , 3 , 4 , 5 , -1 , -1 , 2 , 3] )
print(Q1 , Q2)
```

### Example 3 : Calculate Geometric / Arithmetic mean using *args functions

```python
def Avg( *Inputs , mode) :
    if mode == 'g' :
        return GeoAvg(Inputs)
    elif mode == 'a' :
        return sum(Inputs) / len(Inputs)
    
Q3 = Avg(1 , 2 , 3 , 4 , 1 , 3 , 4 , 6 , 7 , 5 , 5 , 5 , -1 , 2 , 3 , mode = 'g')
print(Q3)
```

### Example 5 : Function to calculate azimuth from two points

```python
def Azimuth(x1 , y1 , x2 , y2) :
    # Calculat the Azimuth from north [ degree ]
    dx = x2 - x1
    dy = y2 - y1
    # Angle to East [ Radian]
    Alpha = np.atan2(dx , dy)

    # Convert to Degree
    DegAzm = np.degrees(Alpha)

    # bring angle in [0 , 360]
    if DegAzm < 0 :
        DegAzm += 360
    
    return DegAzm

print(Azimuth(0 , 0 , 10 , 10)) # 45deg
print(Azimuth(0 , 0 , 0 , 10)) # 0deg
print(Azimuth(0 , 0 , 10 , 0)) # 90deg
```

### Example 9 : Separating points with large errors
We have a list of residual errors for leveling observations. Find indices of errors where absolute value exceeds 2.5 times standard deviation

```python
def LargeErrorChecker(res , STD = 10) :
    LargeErrors = []
    LargeErrorIndices = []
    for i , r in enumerate(res) :
        if abs(r) > (2.5 * STD) :
            LargeErrors.append(r)
            LargeErrorIndices.append(i)
    
    return LargeErrors , LargeErrorIndices

Residuals = [3.2 , -6.5 , 2.1 , 12.3 , -4.0 , 7.8 , -11.2 , 1.5]
StdDev = 2.0 # mm

LargeErrors , LargeErrorIndices = LargeErrorChecker(Residuals , StdDev)
print(f"Indides of large errors :\n{LargeErrorIndices}")
print(f"Large errors :\n{LargeErrors}")
```

### Example 10 : Function to convert DMS (Degrees, Minutes, Seconds) matrix to decimal degrees

```python
def DMS2Decimal(DMSArray) :
    # Convert DMS to Decimal
    # input : an array including numbers with the shape (n , 3) which each row is [d , m , s] form

    Degs = DMSArray[: , 0]
    Mins = DMSArray[: , 1]
    Secs = DMSArray[: , 2]

    Decimal = Degs + Mins/60 + Secs/3600
    return Decimal.reshape(-1 , 1)

DMSC = np.array([[20,12,50],
                 [100,30,2],
                 [342,43,32]])

print(DMSC)
print(DMS2Decimal(DMSC))
```

### Example 11 : Function to calculate distance between points using coordinates and indices

```python
def LEn(x , y , ij):
    I = ij[: , 0]
    J = ij[: , 1]
    # Calculate the substranding in form of vector
    dx = x[J] - x[I]
    dy = y[J] - y[I]
    #calculate the norm ( length ) for each row
    Lengths = np.sqrt(dx**2 + dy**2)
    return Lengths.reshape(-1 , 1) # convert to بردار سویی

XABCD = np.array([0,50,100,40])
YABCD = np.array([0,86.602,0,30])
ij = np.array([[0,2],
               [1,3],
               [2,3]])

# calculations of length
L0 = LEn(XABCD,YABCD,ij)
print(L0)
```

### Recursive Functions in Python
A function that calls itself within its definition.
To prevent infinite loops, there must always be a base condition where the function returns a result without calling itself again.

```python
def Factorial(n) :
    # Calculate n! with بازگشتی method
    if n <= 1 :
        return 1
    else :
        return n* Factorial(n-1)

# test
for i in range(20) :
    x = Factorial(i)
    print(f"Factorial : {i}! = {x}")
```
