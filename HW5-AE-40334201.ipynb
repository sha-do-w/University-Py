## VECTORIZATION

```python
# The BIG ARRAY
ArrBIG = np.random.rand(10000000)

# with Loop
print("LOOP :")
Start = time.time()
MaxVal = ArrBIG[0]
for x in ArrBIG:
    if x > MaxVal:
        MaxVal = x
print(f"Time : {time.time() - Start}\nMax Value = {MaxVal}")

print("=========================")

# with Vectorization
print("VECTORIZATION :")
Start = time.time()
MaxVal = np.max(ArrBIG)
print(f"Time : {time.time() - Start}\nMax Value = {MaxVal}")
```

```python
# Method 1
LongArr = np.arange(1,100001)
print(f"The Array :\n{LongArr}")
StartOne = time.time()
for i , v in enumerate(LongArr) :
    LongArr[i] **= 2

DeltaTimeOne = time.time() - StartOne
print(f"The Powered Array by loop :\n{LongArr}\n\nTime : {DeltaTimeOne}\nShape = {LongArr.shape}")
print("=========================")

# Method 2
LongArr = np.arange(1, 100001)
StartTwo = time.time()
LongArr **= 2
DeltaTimeTwo = time.time() - StartTwo
print(f"The Powered Array by vectorization :\n{LongArr}\n\nTime : {DeltaTimeTwo}\nShape = {LongArr.shape}")
print(f"\nTime optimization : {DeltaTimeOne / DeltaTimeTwo}")
```

```python
# Args FOR NUMEROUS VARIABLES WITHOUT NAME
def SumOne( *Nums ) : # * MAKES THE FUNCTION INPUT CAPACITY UNLIMIT
    result = 0
    for num in Nums :
        result += num
        return result
        # YOU CAN GIVE 1 , 2 , 3 OR EVEN 100 NUMS

print(SumOne(2,4,5))
# print(SumOne(30,20,10))
# print(SumOne(5,4,3,2,1))
```

```python
def FFour(x, y) :
    S = x**2 + y**2
    T = x**2 - y**2
    return S,T,x,y
```

```python
# IF WE WANT TO CONTAIN OUTPUTS OF FUNCTION IN SEPARATE VARIABLES
OutOne , OutTwo , OutThree , OutFour = FFour(132.351, 567.13)
print(OutOne , OutTwo , OutThree , OutFour)
```

```python
# WE SEPARATE ONE VARIABLE FROM OTHERS
OutFive , *OutSix = FFour(132.351, 567.13)
print(OutFive , *OutSix)
```

```python
Out = FFour(132.351, 567.13)
print(Out)
```

```python
OutZ1, *OutZ2 = FFour(0.1, 0.1)
print(OutZ1)
print(OutZ2)
```

# CALCULATOR

```python
def Calculator(): # A SIMPLE CALCULATOR WITH VARIOUS FEATURES
    def Plus(a , b) :
        return a + b
    
    def Minus(a, b) :
        return a - b
    
    def Times(a, b) :
        return a * b
    
    def Divide(a , b) :
        if b == 0 :
            return "ERROR ! DENOMINATOR IS 0 !!!"
        return a / b

    print("====== Calculator ======")
    print("Operations :\nSummation+ ( P )\nSubtraction- ( M )\nMultiplicationx ( T )\nDivision÷ ( D )")
    while True :
        Operation = input("PLEASE CHOSE THE OPERATOR [ EXIT = 'E' ] : >")
        if Operation == 'E' :
            break
        
        if Operation in ['P','M','T','D'] :
            a = float(input("FIRST NUM  : >"))
            b = float(input("SECOND NUM : >:"))
            if Operation == 'P' :
                print(f"RESULT = {Plus(a, b)}\n==========")
            elif Operation == 'M' :
                print(f"RESULT = {Minus(a, b)}\n==========")
            elif Operation == 'T' :
                print(f"RESULT = {Times(a, b)}\n==========")
            elif Operation == 'D' :
                print(f"RESULT = {Divide(a, b)}\n==========")
        else:
            print("INVALID INPUT!\nHINT : PLEASE ENTER THE OPERATOR LETTER CAPITALCASE")

# Run Calc
Calculator()
```
