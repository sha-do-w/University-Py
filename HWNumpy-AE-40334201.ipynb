### PROBLEM 1 : Calculate the distance of points from the base by "Broadcasting"

```python
Stations = np.array([[10,20],[30,40],[50,60],[70,80]])
Base = np.array([35,45])
print(f"The stations :\n{Stations}")
print(f"The base :\n{Base}")

# All in one
Distances = np.sqrt(np.sum(np.pow(Stations-Base,2),axis = 1))

# step by stem equivalent :
# StaMinusBase = Stations-Base
# StaPowerTwo = StaMinusBase**2
# StaSqrt = np.sqrt(np.sum(StaPowerTwo, axis = 1))
# Distances = StaSqrt

print("The Distance of each point from base :")
for i in range(len(Distances)) :
    print(f"Point {i+1} : {Distances[i]}")
# Call the number of the closest point
print(f"\nso the closest point is the \"point {np.argmin(Distances)+1}\"")
```

### PROBLEM 3 : Filteration of the outside points

```python
MinCoordinate = np.array([0,0])
MaxCoordinate = np.array([100,100])

# Generating the point cloud
PointCloud = np.random.uniform(-100, 200, (2000, 2))

# Filteration for insiders : if the points on the contour are in 
MaskXInside = (PointCloud[:,0] >= MinCoordinate[0]) & (PointCloud[:,0] <= MaxCoordinate[0])
MaskYInside = (PointCloud[:,1] >= MinCoordinate[1]) & (PointCloud[:,1] <= MaxCoordinate[1])
# Filteration for outsiders : if the points on the controu are in
# same as : MaskXOutside = (PointCloud[:,0] < MinCoordinate[0]) | (PointCloud[:,0] > MaxCoordinate[0])
MaskXOutside = ~MaskXInside
# same as : MaskYOutside = (PointCloud[:,1] < MinCoordinate[1]) | (PointCloud[:,1] > MaxCoordinate[1])
MaskYOutside = ~MaskYInside

# Create inside / outside points array
InsidePoints = PointCloud[MaskXInside & MaskYInside]
OutsidePoints = PointCloud[MaskXOutside | MaskYOutside]

# Display the data
print(f"The points :\n{PointCloud}")
print("=======================")
print(f"The inside points :\n{InsidePoints}")
print("=======================")
print(f"The outside points :\n{OutsidePoints}")
```

### PROBLEM 6 : Operations on Different Matrices

```python
# Creating matrices ( custom )
Arr1 = np.array([[-4],
              [7],
              [95]])
Arr2 = np.array([[3,-77,17,-314159]])
print(f"The 3x1 array :\n{Arr1}")
print(f"The 1x4 array :\n{Arr2}")

# sum
PlusAB = Arr1 + Arr2
print(f"A + B =\n{PlusAB}\n===================")
print(f"The shape of A + B : {np.shape(PlusAB)}")

# multiply
TimesAB = Arr1 * Arr2
print(f"A * B =\n{TimesAB}")
```

### Description :<br>
Sum : A is (3,1) and B is (1,4) , so "BROADCASTING" aligns them ; by stretching A to 4 columns ( same as B ) and stretching B to 3 rows ( same as A ) with stride tricks.<br><br>
Multiply : Element wise multiplication using "BROADCASTING" ; like sum , stretching A and B and then , multiplying each index ij of A to same ij index at B. that means Aij to Bij , thats it! unlike linear multiplication where we stack up the multiplication of aik to bkj.

An example of "BROADCASTING" for (5,2) and (,2)

```python
# An example of br
FiveTwo = np.array([[43, -22],
                    [-5, -21],
                    np.random.randint(1, 30, size=(1, 2))[0],
                    np.random.randint(-11, 11, size=(1, 2))[0],
                    [-2, 8]])

# if (,2) is (1,2)
IDKTwo1 = np.array([1,2])

# if (,2) is (5,2)
IDKTwo2 = np.array([[45, -22],
                    [-5, -2],
                    np.random.randint(1, 30, size=(1, 2))[0],
                    np.random.randint(-11, 11, size=(1, 2))[0],
                    [-6, 28]])
# display
print(f"if the initial array sums with :\nS -> answer =\n{FiveTwo+IDKTwo1}")
print(f"if the initial array sums with :\nS -> answer =\n{FiveTwo+IDKTwo2}")
```

### PROBLEM 8 : Statistics of an array

```python
# Generate the array
RandomInt = np.random.randint(0,101,(4,5))
print(f"The initial array :\n{RandomInt}\n=================")

# Calculate the avg of each row
RowMeans = np.mean(RandomInt, axis=1)
print(f"The average of each row is :\nRow 1 = {RowMeans[0]}\nRow 2 = {RowMeans[1]}\nRow 3 = {RowMeans[2]}\n=================")

# Calculate the std of each column
ColSTD = np.std(RandomInt, axis=0)
print(f"The standard deviation of each column is :\nRow 1 = {RowMeans[0]}\nRow 2 = {RowMeans[1]}\nRow 3 = {RowMeans[2]}\n=================")

# Extract the biggest value of matrix
MaxVal = np.max(RandomInt)
print(f"The Maximum value of matrix = {MaxVal}\n=================")

# Finding the position [ i , j ] of the lowest value of matrix
MinVal = np.min(RandomInt)
MinIndx = np.where(RandomInt == MinVal)
print("The position(s) of the lowest index(indices) in the matrix :")
for i, (row, col) in enumerate(zip(MinIndx[0], MinIndx[1])):
    print(f"The Index {i+1} : ({row}, {col})")

# Normalize Matrix ( change all element to value of [0,1])
# We had MaxVal and MinVal
NormalMatrix = (RandomInt - MinVal) / (MaxVal - MinVal)
print(f"{NormalMatrix}")
```
