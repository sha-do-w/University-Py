### Example 1

```python
# یک آرایه ی یک بعدی شامل عدد های 1 تا 10 را ایجاد و 

# 1. تمامی عنصر ها را بر 2 تقسیم کنید
# 2. جذر تمامی عنصر ها را محاسبه کنید
# 3. مجموع ، میانگین و انحراف معیار عدد ها را به دست بیاورید

import numpy as np
# Creating Array
Arr = np.array(range(1,11))
print(f"{Arr}\n")

# Dividing Array Elements
DivArr = Arr/2
print(f"{DivArr}\n")

# Rooting Array Elements
SqrtArr = np.sqrt(Arr)
print(f"{SqrtArr}\n")

# Calculating Sum of Array Elements
SumArr = np.sum(Arr)
print(f"{SumArr}\n")

# Calculating Mean of Array Elements
MeanArr = np.mean(Arr)
print(f"{MeanArr}\n")

# Calculating Standard Deviation of Array Elements
StdArr = np.std(Arr)
print(f"{StdArr}\n")
```

### Example 4

```python
import numpy as np

Matrix = np.random.randint(1,21,size=(4,4))
print(f"{Matrix}\n")

# Avg of each Col
ColMeans = np.mean(Matrix, axis=0)
print(f"{ColMeans}\n")

# Sum of each Row
RowSum = np.sum(Matrix,axis=1)
print(f"{RowSum}\n")

# Min of each Col
ColMin = np.min(Matrix,axis=0)
print(f"{ColMin}\n")

# Max of Matrix
MaxIndex = np.max(Matrix)
print(f"{MaxIndex}\n")

# Normalize Matrix ( change all element to value of [0,1])
# we had MaxIndex
MinIndex = np.min(Matrix)
NormalMatrix = (Matrix - MinIndex) / (MaxIndex - MinIndex)
print(f"{NormalMatrix}")
```

### Attributes

```python
import numpy as np

# 0d
Arr0 = np.array(-2.5)

# 1d
Arr1 = np.array([-14, 22, 3, 4])

# 2d
Arr2 = np.array([[1,2],
                 [3,4]])

# 3d
Arr3 = np.array([[[1,2],
                  [3,4]],
                  [[5,6],
                   [7,8]]])

# s and e
s = np.array([[1],[2],[3]])

e = np.array([[-1],
              [ 2],
              [ 4]])

# .ndim , .shape , .size
print("\n===== ndim shape size =====")
print(Arr0.ndim , Arr1.ndim , Arr2.ndim , Arr3.ndim , s.ndim , e.ndim)
print(Arr0.size , Arr1.size , Arr2.size , Arr3.size , s.size , e.size)
print(Arr0.shape , Arr1.shape , Arr2.shape , Arr3.shape , s.shape , e.shape)

# .dtype
print("\n===== dtype =====")
print(Arr0.dtype , Arr1.dtype , Arr2.dtype , Arr3.dtype , s.dtype , e.dtype)

# Methods
print("\n===== Methods =====")
print(Arr2.min()) # 1st way
print(np.min(Arr2)) # 2nd way

```

### Data type & Conversion

```python
import numpy as np

# Data Type
Arr11 = np.array([1 ,2,3,4])
Arr22 = np.array([1.,2,3,4])

print(Arr11.dtype, Arr22.dtype)
print(Arr11,Arr22)

# Define array with target data type
print("\n===== Define data type =====")
Arr1 = np.array([130,128,127,400] , dtype = np.int16)
Arr2 = np.array([1,2,255,20] , dtype = np.uint8)
print(Arr1 , Arr1.dtype , Arr2 , Arr2.dtype)

Arr3 = np.array([0,255,256,-1,-2] , dtype = np.int32)
print(Arr3 , Arr3.dtype)

Arr4 = np.array([complex(1,1),128,127,400] , dtype = np.complex64)
print(Arr4 , Arr4.dtype)

Arr5 = np.array([130,0,127,400] , np.bool_)
print(Arr5 , Arr5.dtype)

# Convert
print("\n===== Convert =====")
Arr6 = np.array([130,128,127,400] , dtype = np.int32)
print(Arr6 , Arr6.dtype)
Arr6 = Arr6.astype(np.float64)
print(Arr6.dtype)
```

### Special Arrays

```python
import numpy as np

# np.zeros()
print("\n===== Zero =====")
ArrZero2d = np.zeros((3,4) , dtype = np.int16)
print(ArrZero2d)
ArrZero3d = np.zeros((5,6,4) , dtype = np.int16)
print(ArrZero3d)

# np.ones()
print("\n===== One =====")
ArrOne1d = np.ones(6)
ArrOne2d = np.ones((4,5) , dtype = np.float32)
print(ArrOne1d , ArrOne2d , ArrOne2d.dtype)

# np.eye()
print("\n===== I =====")
ArrI2d = np.eye(3,2)
ArrI1d = np.eye(3 , dtype = np.int8)
print(f"{ArrI2d}\n\n{ArrI1d}")

# np.empty()
print("\n===== Empty =====")
print(np.empty(5))

# np.full()
print("\n===== Full =====")
ArrFull1 = np.full((3,2,2) , 1.2)
ArrFull2 = np.ones((3,2,2)) * 1.2
print(f"{ArrFull1}\n\n{ArrFull2}")

print("\n===== Arange =====")
PyAr = np.array(range(2,10,2))
NpAr = np.arange(2,10,2)
print(PyAr , NpAr)

print("\n===== Linspace =====")
LinArr = np.linspace(0,9,10)
LinArr50 = np.linspace(0,9,50)
ArArr = np.arange(0,10,1)
print(LinArr , LinArr50 , ArArr)

print("\n===== Step =====")
S1 = np.linspace(0,1,100)
S2 = np.linspace(0,1)
print(f"{S1}\n{S2}")
```

### Create a Zero Vector with the same size as A

```python
import numpy as np

A = np.zeros((5,6,4) , dtype = np.int16)
print(A.shape , A.dtype)

# Hard way
print("\n===== Hard =====")
B1 = np.zeros(A.shape , dtype = A.dtype)
print(B1.shape , B1.dtype)

# Easy way
print("\n===== Easy =====")
B2 = np.zeros_like(A)
print(B2.shape , B2.dtype)
print("\n===== Ones =====")
B3 = np.ones_like(A)
print(B3.shape , B3.dtype)
```
