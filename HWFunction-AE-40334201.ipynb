### PROBLEM 1 : Convert Polar to Decart and Vice Versa

```python
import numpy as np
def P2D (UserP) :
    """
    Convert Polar to Decart coordinates.
    
    Input: 2D array [[Radius , Angle], ...] where angle in degrees or radians
    Output: (x , y)
    
    Example:
    >>> P2D(np.array([[5 , 30]]))  # Radius = 5 , Angle = 30 degrees
    (array([4.33]), array([2.5]))
    """
    # x = rCos[theta] , y = rSin[theta]
    if np.any(UserP[:,0] < 0)  :
        return "Math is not Mathing! Negative Radius? Seriously?!"
    while True :
        UserAnswer = input("What kind of angle ? a. Degree\tb. Radian\n\t>>")
        match(UserAnswer) :
            case "a" :
                x = UserP[:,0] * np.cos(np.radians(UserP[:,1]))
                y = UserP[:,0] * np.sin(np.radians(UserP[:,1]))
                return np.column_stack((np.round(x,10),np.round(y,10)))
            case "b" :
                x = UserP[:,0] * np.cos(UserP[:,1])
                y = UserP[:,0] * np.sin(UserP[:,1])
                return np.column_stack((np.round(x,10),np.round(y,10)))
            case _ :
                print("ERROR! INVALID VALUE!")

def D2P(UserP) :
    """
    Convert Decart to Polar coordinates.
    
    Input : 2D array [[x , y], ...]
    Output : (Radius , Angle)
    
    Example :
    >>> P2D(np.array([[2 , -2]]))  # x = 2 , y = -2
    (array([2 . sqrt(2)]), array([-45])) -> deg
    (array([2 . sqrt(2)]), array([sqrt(2) / 2]) -> rad
    """
    UserX = UserP[:,0]
    UserY = UserP[:,1]
    # radius = (x^2 + y^2)^0.5 , angle = ArcTan(y/x)
    while True :
        UserAnswer = input("What kind of angle ? a. Degree\tb. Radian\n\t>>")
        match(UserAnswer) :
            case "a" :
                Radius = np.sqrt(np.pow(UserX,2) + np.pow(UserY,2))
                Angle = np.degrees(np.atan2(UserY , UserX))
                return np.column_stack((Radius , Angle))
            case "b" :
                Radius = np.sqrt(np.pow(UserX,2) + np.pow(UserY,2))
                Angle = np.atan2(UserY , UserX)
                return np.column_stack((Radius , Angle))
            case _ :
                print("ERROR! INVALID VALUE!")
 

UserNum = np.array([[2,-32],[8,92],[5,-12],[7,50],[50,90]])
Converted = P2D(UserNum)
ConvertedBack = D2P(Converted)
print(f"At first , it was polar :\n{UserNum}\nThen , it converted to decart :\n{Converted}\nAnd then , it converted back to polar :\n{ConvertedBack}")
```

### PROBLEM 3 : Area of a polygon

```python
import numpy as np

def ListPLArea(UserP) :
    Area = 0
    UserP.append(UserP[0])
    for i in range(len(UserP) - 1) :
        x1 , y1 = UserP[i]
        x2 , y2 = UserP[i+1]
        Area += abs((x1 * y2) - (x2 * y1))
    return 0.5 * Area

def ArrayPLArea(UserP) :
    UserP = np.append(UserP,UserP[0,:].reshape(1,-1),axis = 0)
    PtX = UserP[:,0]
    PtY = UserP[:,1]
    XinY = PtX[0:-1] * PtY[1:]
    YinX = PtY[0:-1] * PtX[1:]
    Area = np.sum(0.5 * np.abs(XinY - YinX))
    return Area

UserP1 = np.array([[-2,-2],[-2,2],[2,2],[2,-2]])
UserP2 = [[0,0],[0,4],[4,4],[4,0]]
Area1 = ArrayPLArea(UserP1)
Area2 = ListPLArea(UserP2)
print(Area1)
print(Area2)
```

### EXAMPLE 4 : GPS Stations Data Management by functions

```python
import numpy as np

def LPXYZDist(IDs , X , Y , Z) :
    Distances = np.zeros((len(IDs) , len(IDs)))
    for i in range(len(IDs)) :
        for j in range(len(IDs)) :
            Distances[i , j] = np.sqrt((X[i] - X[j])**2 + (Y[i] - Y[j])**2 + (Z[i] - Z[j])**2)
    return Distances

def BRXYZDist(IDs , X , Y , Z , Pt = None) :
    # Reshape to column vectors
    XColumn = np.array(X).reshape(-1 , 1)
    YColumn = np.array(Y).reshape(-1, 1)
    ZColumn = np.array(Z).reshape(-1, 1)

    # Broadcasting does the rest
    Distances = np.sqrt((XColumn - X)**2 + (YColumn - Y)**2 + (ZColumn - Z)**2)
    
    if Pt is not None:
        Pt = np.array(Pt)
        Pts = np.array([X , Y , Z]).T
        Dx = Pts - Pt
        DistFromPt = np.sqrt(np.sum(Dx**2 , axis=1)) 
        return Distances , np.min(DistFromPt)
    else:
        return Distances
IDs = [1,2,3,4]
X = [0,9,4,1]
Y = [10,34,2,-4]
Z = [6,-3,100,-200]
print(LPXYZDist(IDs , X , Y , Z))
print("+================")
print(BRXYZDist(IDs , X , Y , Z))
print("+================")
print(BRXYZDist(IDs , X , Y , Z , Pt = np.array([-9,6,-3])))
```

### EXAMPLE 6 : Prime number

```python
import numpy as np
def IsPrime(num) :
    match(num) :
        case num if num < 0 :
            return "Negative ?! Really ?!"
        case num if (num == 0) | (num == 1) :
            return f"Wrong key ? or lack of knowledge ? because {num} is no either Prime or Composite !"
        case num if num == 2 :
            return "Absolutely ! It is the only Even Prime number !"
        case num if num % 2 == 0 :
            return "If you want to ensure the code works fine , it works fine !"
        case num if num >= 3 :
            for i in range(3 , int(pow(num , 13/26)) + 1, 2) :
                if num % i == 0 :
                    return "Your number IS NOT Prime !"
            return "Your number IS Prime !"
        case _ :
            return "Error ! Invalid Value"
        
def WhichIsPrime(NumOne , NumTwo) :
    UserP = range(NumOne , NumTwo + 1)
    PrimeNums = []
    for i in range(len(UserP)) :
        con = IsPrime(UserP[i])
        if con == "Your number IS Prime !" or con == "Absolutely ! It is the only Even Prime number !" :
            PrimeNums.append(UserP[i])
    return PrimeNums
            

print(IsPrime(-3))
print(IsPrime(0))
print(IsPrime(2))
print(IsPrime(3))
print(IsPrime(6))
print(IsPrime(9))
print(WhichIsPrime(5,85))
```

### EXAMPLE 7 : Fibonacci sequence

```python
import numpy as np
def Fibonacci(NumOfItr) :
    # We used to solve it in Linear Algebra lecture
    DesignMatrix = np.array([[1, 1],[1, 0]])
    AOne = np.array([1, 0])
    return (np.linalg.matrix_power(DesignMatrix, NumOfItr) @ AOne)[1]

print(Fibonacci(20))
```

### EXAMPLE 10 : Validate input

```python
import numpy as np
def ValidateNums(UserP = np.empty(1)) :
    UserP = np.array(UserP)
    if np.all(UserP > 0) :
        return np.sum(UserP)
    elif np.any(UserP <= 0) :
        return "Warning ! There is at least one non-positive number !"
    elif np.isnan(UserP).all() :
        print("The input is empty !")
        return 0

a = ValidateNums(np.array([1,2,3,4,5,5]))
b = ValidateNums(np.array([1,-2,3,-4,5,5]))
c = ValidateNums(np.array([]))
print(a)
print(b)
print(c)
```
