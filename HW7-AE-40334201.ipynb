### System of Equations
$$L = AX -> L_{m\times 1} = A_{m\times n} X_{n\times 1}$$
For systems with the condition $n < m$ :
$$Rank_{max}[A] = n < m$$
$$df = m - n$$
Least square method : finds the best estimate by minimizing the norm of the residual error vector.
$$\hat x = (A^T A)^{-1} A^T L$$
Weight matrix : $W_{m\times m}$
$$\hat x = (A^T WA)^{-1} A^T WL$$
The Best Linear Unbiased Estimator (BLUE) :
$$W = Q_L^{-1} -> \hat x = (A^T Q_L^{-1}A)^-1 A^T Q_L^{-1}L

### EXAMPLE 1 : Create the W and Cov matrices

```python
def CovWeightMatrix(Vars) :
    Cov = np.diag(Vars)
    Weight = np.linalg.inv(Cov)
    return Cov , Weight

# Variance of Observations
Variances = np.array([2,3,4]) # SIGMA^2 -> mm^2

# Diagonal Cov matrix & Weight matrix [ the inverse of Diagonal Cov ]
Cov , Weight = CovWeightMatrix(Variances)
print(f"Cov matrix :\n{Cov}")
print(f"Weight matrix :\n{Weight}")
```

### EXAMPLE 2 : Solve the System of Equations
the L = AX -> np.linalg.solve()

```python
# +6x - 2y = +10
# -2x + 4y = -4
# ===================================
# Defining the A and L and solve X
A = np.array([[6,-2],
              [-2,4]])
L = np.array([[10],
              [-4]])
X = np.linalg.solve(A,L)
print(f"The Unknowns [numpy] :\n{X}")

# Using xHat formula
ATA = A.T @ A
ATL = A.T @ L
X = np.linalg.inv(ATA) @ ATL

# Display the result
print(f"X =\n{X}")
```

### EXAMPLE 3 : Solve with Least Square

```python
# Defining the A and L and solve X
A = np.array([[1, 0],
              [1, 1],
              [0, 1]])
L = np.array([[100.2],
              [101.4],
              [99.8]])

# Using xHat formula
ATA = A.T @ A
ATL = A.T @ L
X = np.linalg.inv(ATA) @ ATL

# Residual vector
V = A @ X - L

# Display the result
print(f"ATA :\n{ATA}\n")
print(f"ATL :\n{ATL}\n")
print(f"X Unknowns :\n{X}\n")
print(f"V Residuals :\n{V}\n")
```

## PARAMETRIC MODEL
if we want to write the Observations based on the Unknowns , this model is called "Parametric model" :
<br><br>
$$L = F(x) \xrightarrow{\text{ Derivate }} Coefficients$$
<br><br>
$$Coefficients \xrightarrow{\text{    Matrix    }} A$$
<br><br>
$$\hat x = (A^T PA)^{-1} A^T PL$$
<br><br>
$$\sigma _{\hat x\hat x} = \sigma _0 ^2 (A^T PA)^{-1}$$
<br><br>
$$\hat v = A\hat x - L$$
<br><br>
$$\sigma _{\hat v\hat v} = \sigma _0 ^2 (P^{-1} - A(A^T PA)^{-1} A^T)$$
<br><br>
$$\hat L = L + \hat v$$
<br><br>
$$\sigma _{\hat L\hat L} = \sigma _0 ^2 A(A^T PA)^{-1} A^T$$

### EXAMPLE 5 : We want to calculate the Height of all points in a Leveling network with 1 known point [ = 100m ]

$$H_1 = 100m$$
$$\Delta H_{12} = 5.430m$$
$$\Delta H_{23} = 2.380m$$
$$\Delta H_{31} = 7.790m$$

```python
H1 = 100
DH12 = 5.430
DH23 = 2.380
DH31 = 7.790

# DH12 + H1 = H2
# DH23 = H3 - H2
# DH31 - H1 = H3

A = np.array([[1,0],
              [-1,1]
              [0,-1]])

L = np.array([DH12 + H1 , DH23 , DH31 - H1])
LColumn = L.reshape(-1,1)

ATA = A.T @ A
ATL = A.T @ L
XHat1 = np.linalg.inv(ATA) @ ATL

print(f"L :\n{LColumn}\nXHat :{XHat1}\n")
XHat2 = np.linalg.lstsq(A,L,rcond = 0)
print(f"XHat 2 :\n{XHat2}")
```

### EXAMPLE 6 : In a Leveling network , 2 points have known heights. Calculate the height of other points
$$H_4 = 100.00m$$
$$H_5 = 100.28m$$
$$\Delta H_{41} = +1.30m$$
$$\Delta H_{15} = -1.00m$$
$$\Delta H_{53} = -0.27m$$
$$\Delta H_{31} = -1.30m$$
$$\Delta H_{52} = -0.50m$$
$$\Delta H_{23} = +0.26m$$
$$\sigma = 2mm = 0.002m$$
<p align="center">
  <img src="hw7-p2.png" width="500"/>
</p>


```python
# from scipy.linalg import inv

# initial data
H4 = 100
H5 = 100.28

DH = np.array([1.3,-1,-0.27,-1.3,-0.5,0.26]).reshape(-1,1)
DH41 = 1.3   # DH41 = H1 - H4 -> DH41 + H4 = +H1
DH15 = -1.0  # DH15 = H5 - H1 -> DH15 - H5 = -H1
DH53 = -0.27 # DH53 = H3 - H5 -> DH53 + H5 = H3
DH31 = -1.3
DH52 = -0.5
DH23 = 0.26
# Numerical L
L = np.array([DH[0] + H4 , DH[1] - H5 , DH[2] + H5 , DH[3] , DH[4] + DH[5] , DH[5]]).reshape(-1,1)

print(f"L :\n{L}")
Sigma = 0.002 # 2mm = 0.2cm = 0.002m
QL = np.pow(Sigma,2) * np.eye(6)

# Symbolic Variables
H1 , H2 , H3 = sp.symbols('H1 H2 H3')

# Define L1 to L6 symbolic
L1 = H1
L2 = -H1
L3 = H3
L4 = H1 - H3
L5 = H2
L6 = H3 - H2

# Symbolic A Matrix
ASym = sp.zeros(0,3)
ASym[0,0] = sp.diff(L1,H1)
ASym[0,1] = sp.diff(L1,H2)
ASym[0,2] = sp.diff(L1,H3)

ASym[1,0] = sp.diff(L2,H1)
ASym[1,1] = sp.diff(L2,H2)
ASym[1,2] = sp.diff(L2,H3)

ASym[2,0] = sp.diff(L3,H1)
ASym[2,1] = sp.diff(L3,H2)
ASym[2,2] = sp.diff(L3,H3)

ASym[3,0] = sp.diff(L4,H1)
ASym[3,1] = sp.diff(L4,H2)
ASym[3,2] = sp.diff(L4,H3)

ASym[4,0] = sp.diff(L5,H1)
ASym[4,1] = sp.diff(L5,H2)
ASym[4,2] = sp.diff(L5,H3)

ASym[5,0] = sp.diff(L6,H1)
ASym[5,1] = sp.diff(L6,H2)
ASym[5,2] = sp.diff(L6,H3)

# Transform Symbolic Matrix to Numerical [ because every ij is constant ]
ANum = np.array(ASym , dtype = np.float64)
print(f"ANum :\n{ANum}")

# Calculate estimated parameters with LeastSquare
W = np.linalg.inv(QL)
print(f"W :\n{W}")
ATWA = ANum.T @ W @ ANum
ATWL = ANum.T @ W @ L
XHat = np.linalg.inv(ATWA) @ ATWL
LHat = ANum @ XHat
eHat = L - LHat

# Display Result
print(f"Observations :\n{L.flatten}")
print(f"Matrix A [ Numerical ] :\n{ANum}")
print(f"Estimated Parameters [ xHat ] :\n{XHat.flatten}")
print(f"Predict the L [ lHat ] :\n{LHat.flatten}")
print(f"Residues [ eHat ] :\n{eHat.flatten}")
```
